# Alfido Tech Employee Attrition Analysis

EDA, KPI analysis, segment comparisons and leadership recommendations.

```
import pandas as pd
import numpy as np
raw = pd.read_csv('employee_attrition (1).csv')
raw.shape
```

```
raw.isna().sum().sort_values(ascending=False).head(10)
```

```
clean = raw.copy()
for c in clean.select_dtypes(include=np.number).columns:
    if clean[c].isna().any(): clean[c] = clean[c].fillna(clean[c].median())
for c in clean.select_dtypes(include='object').columns:
    if clean[c].isna().any(): clean[c] = clean[c].fillna(clean[c].mode()[0])
clean['AttritionFlag']=(clean['Attrition']=='Yes').astype(int)
clean['AgeBand']=pd.cut(clean['Age'], [17,24,29,34,39,44,49,100], labels=['18-24','25-29','30-34','35-39','40-44','45-49','50+'])
clean['TenureBand']=pd.cut(clean['YearsAtCompany'], [-1,1,3,5,10,100], labels=['0-1','2-3','4-5','6-10','11+'])
```

```
clean.groupby('OverTime')['AttritionFlag'].agg(['mean','count','sum']).assign(attrition_rate_pct=lambda x:x['mean']*100)
```

```
clean.groupby('JobRole')['AttritionFlag'].agg(['mean','count','sum']).assign(attrition_rate_pct=lambda x:x['mean']*100).sort_values('attrition_rate_pct',ascending=False)
```

```
clean.groupby('TenureBand', observed=False)['AttritionFlag'].agg(['mean','count','sum']).assign(attrition_rate_pct=lambda x:x['mean']*100)
```

## Key recommendations
1. Address overtime hotspots.
2. Strengthen first-year onboarding and manager support.
3. Build early-career development pathways.
4. Investigate high-risk roles, especially Sales Representative.
5. Review travel and low-satisfaction pressure points.
